```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Physical Constants
# ------------------------------------------------------------

hbar = 1.054571817e-34       # Reduced Planck constant (J.s)
electron_mass = 9.1093837e-31 # Electron mass (kg)
eV = 1.602176634e-19         # 1 eV in joules


# ------------------------------------------------------------
# 2. User Input
# ------------------------------------------------------------

print("=" * 60)
print("        QUANTUM TUNNELING PROBLEM SOLVER")
print("=" * 60)

print("\nSelect particle:")
print("1. Electron")
print("2. Proton")
print("3. Custom particle")

# Using default values for demonstration
choice = 1 # Electron
print(f"\nEnter choice (1/2/3): {choice} (Defaulting to Electron)")

if choice == 1:
    particle_name = "Electron"
    mass = electron_mass

elif choice == 2:
    particle_name = "Proton"
    mass = 1.6726219e-27

elif choice == 3:
    particle_name = "Custom Particle"
    mass = 9.1093837e-31 # Default custom particle mass to electron mass
    print(f"Enter particle mass (kg): {mass} (Defaulting to Electron Mass)")

else:
    print("Invalid choice!")
    raise SystemExit


# ------------------------------------------------------------
# 3. Problem Parameters
# ------------------------------------------------------------

# Using default values for demonstration
energy_eV = 1.0 # eV
print(f"Enter particle energy (eV): {energy_eV} (Defaulting to 1.0 eV)")
banner_eV = 2.0 # eV
print(f"Enter barrier height (eV): {banner_eV} (Defaulting to 2.0 eV)")
width_nm = 0.5 # nm
print(f"Enter barrier width (nm): {width_nm} (Defaulting to 0.5 nm)")

# Convert units
E = energy_eV * eV
V0 = banner_eV * eV
width = width_nm * 1e-9


# ------------------------------------------------------------
# 4. Check Tunneling Condition
# ------------------------------------------------------------

if energy_eV >= banner_eV:

    print("\nParticle energy is greater than or equal to")
    print("the barrier height.")
    print("Classical tunneling condition E < V0 is not satisfied.")

    raise SystemExit


# ------------------------------------------------------------
# 5. Calculate Decay Constant
# ------------------------------------------------------------

kappa = np.sqrt(2 * mass * (V0 - E)) / hbar


# ------------------------------------------------------------
# 6. Calculate Transmission Probability
# ------------------------------------------------------------

# Thick-barrier approximation
T = np.exp(-2 * kappa * width)

# Reflection probability
R = 1 - T


# ------------------------------------------------------------
# 7. Calculate Wave Numbers
# ------------------------------------------------------------

k = np.sqrt(2 * mass * E) / hbar


# ------------------------------------------------------------
# 8. Display Results
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("                 RESULTS")
print("=" * 60)

print(f"Particle                 : {particle_name}")
print(f"Particle mass            : {mass:.4e} kg")
print(f"Particle energy          : {energy_eV:.3f} eV")
print(f"Barrier height           : {banner_eV:.3f} eV")
print(f"Barrier width            : {width_nm:.3f} nm")

print("-" * 60)

print(f"Wave number (k)          : {k:.4e} 1/m")
print(f"Decay constant (κ)       : {kappa:.4e} 1/m")

print("-" * 60)

print(f"Transmission probability : {T:.6e}")
print(f"Transmission percentage  : {T * 100:.6e} %")

print(f"Reflection probability   : {R:.6e}")
print(f"Reflection percentage    : {R * 100:.6f} %")

print("=" * 60)


# ------------------------------------------------------------
# 9. Create Position Array
# ------------------------------------------------------------

# Barrier starts at x = 0
# Barrier ends at x = width

x_before = np.linspace(-3 * width, 0, 300)
x_barrier = np.linspace(0, width, 300)
x_after = np.linspace(width, 4 * width, 400)


# ------------------------------------------------------------
# 10. Wavefunction
# ------------------------------------------------------------

# Normalize incoming wave amplitude
A = 1.0

# Before barrier
psi_before = A * np.exp(1j * k * x_before)

# Inside barrier
psi_barrier = A * np.exp(-kappa * x_barrier)

# After barrier
psi_after = A * np.sqrt(T) * np.exp(
    1j * k * (x_after - width)
)


# ------------------------------------------------------------
# 11. Combine Arrays
# ------------------------------------------------------------

x = np.concatenate([
    x_before,
    x_barrier,
    x_after
])

psi = np.concatenate([
    psi_before,
    psi_barrier,
    psi_after
])

probability_density = np.abs(psi)**2


# Convert x to nm for plotting
x_nm = x * 1e9


# ------------------------------------------------------------
# 12. Plot Potential Barrier
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(x_nm, np.zeros_like(x_nm),
         label="Particle Energy")

plt.plot(
    [0, width_nm],
    [banner_eV, banner_eV],
    linewidth=3,
    label="Potential Barrier"
)

plt.axvspan(
    0,
    width_nm,
    alpha=0.2
)

plt.xlabel("Position (nm)")
plt.ylabel("Energy (eV)")
plt.title("Quantum Tunneling Potential Barrier")
plt.legend()
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# 13. Plot Wavefunction
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    x_nm,
    np.real(psi),
    label="Wavefunction ψ(x)"
)

plt.axvspan(
    0,
    width_nm,
    alpha=0.2,
    label="Barrier Region"
)

plt.xlabel("Position (nm)")
plt.ylabel("Wavefunction")
plt.title("Quantum Wavefunction During Tunneling")
plt.legend()
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# 14. Plot Probability Density
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    x_nm,
    probability_density,
    label="Probability Density |ψ|²"
)

plt.axvspan(
    0,
    width_nm,
    alpha=0.2,
    label="Barrier Region"
)

plt.xlabel("Position (nm)")
plt.ylabel("|ψ|²")
plt.title("Quantum Probability Density")
plt.legend()
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# 15. Transmission vs Barrier Width
# ------------------------------------------------------------

width_range_nm = np.linspace(0.1, 5, 200)
width_range_m = width_range_nm * 1e-9

T_width = np.exp(
    -2 * kappa * width_range_m
)

plt.figure(figsize=(10, 5))

plt.plot(
    width_range_nm,
    T_width * 100
)

plt.xlabel("Barrier Width (nm)")
plt.ylabel("Transmission Probability (%)")
plt.title("Transmission Probability vs Barrier Width")
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# 16. Transmission vs Particle Energy
# ------------------------------------------------------------

energy_range_eV = np.linspace(
    0.1,
    banner_eV * 0.99,
    200
)

energy_range_J = energy_range_eV * eV

kappa_energy = np.sqrt(
    2 * mass * (V0 - energy_range_J)
) / hbar

T_energy = np.exp(
    -2 * kappa_energy * width
)

plt.figure(figsize=(10, 5))

plt.plot(
    energy_range_eV,
    T_energy * 100
)

plt.xlabel("Particle Energy (eV)")
plt.ylabel("Transmission Probability (%)")
plt.title("Transmission Probability vs Particle Energy")
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# 17. Final Interpretation
# ------------------------------------------------------------

print("\nINTERPRETATION")
print("-" * 60)

if T > 0.01:
    print("Significant quantum tunneling is possible.")

elif T > 1e-6:
    print("Quantum tunneling occurs with a small probability.")

else:
    print("Quantum tunneling probability is extremely small.")

print("\nThe simulation demonstrates that tunneling probability")
print("decreases exponentially as the barrier becomes wider or higher.")
```

# Quantum Tunneling Problem Solver

This notebook calculates and visualizes quantum mechanical tunneling through a 1D rectangular potential energy barrier.

## Physical Model:
- **Region I ($x < 0$):** Free particle wave function $\psi_I(x) = A e^{i k x} + B e^{-i k x}$
- **Region II ($0 \le x \le L$):** Evanescent wave function $\psi_{II}(x) = C e^{-\kappa x} + D e^{\kappa x}$
- **Region III ($x > L$):** Transmitted wave function $\psi_{III}(x) = F e^{i k x}$

where $k = \frac{\sqrt{2mE}}{\hbar}$ and $\kappa = \frac{\sqrt{2m(V_0-E)}}{\hbar}$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
hbar = 1.054571817e-34  # J*s
e_charge = 1.602176634e-19  # C
m_e = 9.10938356e-31  # kg

# Input parameters
particle = 'Electron'
E_eV = 1.0
V0_eV = 2.0
L_nm = 0.5

# Conversions
E = E_eV * e_charge
V0 = V0_eV * e_charge
L = L_nm * 1e-9

# Wave vector and decay constant
k = np.sqrt(2 * m_e * E) / hbar
kappa = np.sqrt(2 * m_e * (V0 - E)) / hbar if V0 > E else 0.0

# Transmission & Reflection Probabilities
if E < V0:
    T = 1.0 / (1.0 + (V0**2 * np.sinh(kappa * L)**2) / (4.0 * E * (V0 - E)))
elif E > V0:
    k2 = np.sqrt(2 * m_e * (E - V0)) / hbar
    T = 1.0 / (1.0 + (V0**2 * np.sin(k2 * L)**2) / (4.0 * E * (E - V0)))
else:
    T = 1.0 / (1.0 + (m_e * V0 * L**2) / (2.0 * hbar**2))

R = 1.0 - T

print('=' * 60)
print('        QUANTUM TUNNELING PROBLEM SOLVER')
print('=' * 60)
print(f'Particle                 : {particle}')
print(f'Particle mass            : {m_e:.4e} kg')
print(f'Particle energy          : {E_eV:.3f} eV')
print(f'Barrier height           : {V0_eV:.3f} eV')
print(f'Barrier width            : {L_nm:.3f} nm')
print('-' * 60)
print(f'Wave number (k)          : {k:.4e} 1/m')
print(f'Decay constant (κ)       : {kappa:.4e} 1/m')
print('-' * 60)
print(f'Transmission probability : {T:.6e}')
print(f'Transmission percentage  : {T * 100:.6f} %')
print(f'Reflection probability   : {R:.6e}')
print(f'Reflection percentage    : {R * 100:.6f} %')
print('=' * 60)
```

```python
# Figure 1: Wavefunction & Potential Energy Diagram
plt.figure(figsize=(10, 5))
x_nm = np.linspace(-1.0, 1.5, 1000)
x = x_nm * 1e-9
V_x = np.where((x >= 0) & (x <= L), V0_eV, 0.0)

plt.plot(x_nm, V_x, 'r-', linewidth=2, label='Potential Barrier V(x)')
plt.axhline(E_eV, color='b', linestyle='--', label='Particle Energy E')
plt.axvspan(0, L_nm, color='red', alpha=0.1, label='Barrier Region')

psi = np.zeros_like(x_nm)
reg1 = x_nm < 0
reg2 = (x_nm >= 0) & (x_nm <= L_nm)
reg3 = x_nm > L_nm

psi[reg1] = E_eV + 0.2 * np.cos(k * x[reg1])
psi[reg2] = E_eV + 0.2 * np.exp(-kappa * x[reg2])
psi[reg3] = E_eV + 0.2 * np.sqrt(T) * np.cos(k * (x[reg3] - L))

plt.plot(x_nm, psi, 'g-', label='Wavefunction ψ(x)')
plt.xlabel('Position x (nm)')
plt.ylabel('Energy (eV)')
plt.title('Quantum Tunneling: Potential Barrier & Wavefunction')
plt.legend(loc='upper right')
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Figure 2: Transmission Probability vs. Barrier Width
plt.figure(figsize=(10, 5))
L_range_nm = np.linspace(0.01, 2.0, 500)
L_range = L_range_nm * 1e-9

T_list = [1.0 / (1.0 + (V0**2 * np.sinh(kappa * l_val)**2) / (4.0 * E * (V0 - E))) for l_val in L_range]

plt.semilogy(L_range_nm, T_list, 'b-', linewidth=2, label=f'Transmission T (E={E_eV} eV, V0={V0_eV} eV)')
plt.axvline(L_nm, color='k', linestyle='--', label=f'Current L = {L_nm} nm (T = {T:.4e})')
plt.xlabel('Barrier Width L (nm)')
plt.ylabel('Transmission Probability T (log scale)')
plt.title('Transmission Probability vs. Barrier Width')
plt.legend(loc='upper right')
plt.grid(True, which='both', linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```
